# Efficiency wages with an informal sector

An extension of Shapiro & Stiglitz (1984), *Equilibrium Unemployment as a Worker
Discipline Device*, to a dual labour market.

**Question:** how does an informal sector — a productive fallback for fired workers,
rather than a static unemployment benefit — change the no-shirking condition, the
equilibrium wage and the welfare properties of an efficiency-wage economy?

**Approach:** analytical equilibrium by root-finding, then an agent-based simulation
with heterogeneous workers and firms, shock experiments (including an AI-style
multi-shock), and a calibration to Colombian labour-market data.

## 1. The paper

The starting point for this project is the seminal paper: "Equilibrium Unemployment as a Worker Discipline Device" by Carl Shapiro and Joseph E. Stiglitz, published in The American Economic Review, Vol. 74, No. 3 (June, 1984).


**Mathematical Framework and the model:**


The economy consists of $N$ identical workers who maximize their expected lifetime utility. At any moment, a worker is either employed or unemployed.  Utility Function: $U = w - e$. Workers value wages ($w$) and dislike effort ($e$).  The Discrete Choice: A worker chooses either to provide effort ($e > 0$) or to shirk ($e = 0$). While shirking provides an immediate utility gain (since $e=0$) but introduces a probability $q$ per unit of time of being caught and fired.

We use the following equations to value the states of employment. 
The value of a job ($V_E$) is seen as an asset that pays "dividends" (utility) and can change in value (if you lose the job).

1. **Value of a Non-Shirker ($V_E^N$): $$rV_E^N = (w - e) + b(V_u - V_E^N)$$**

The interest on the asset ($rV_E^N$) equals the flow of utility $(w-e)$ minus the risk of losing the job due to exogenous reasons ($b$).

2. **Value of a Shirker ($V_E^S$): $$rV_E^S = w + (b + q)(V_u - V_E^S)$$**

A shirker gets a higher flow (no $-e$) but faces a higher probability of job loss $(b+q)$ because they might be caught.  

3. **Value of the Unemployed ($V_u$):** $$rV_u = \bar{w} + a(V_E - V_u)$$

An unemployed person gets benefits $\bar{w}$ and has a probability $a$ of finding a job

**Therefore:**

To ensure workers don't shirk, firms must pay a wage such that $V_E^N \ge V_E^S$. By setting $V_E^N = V_E^S$ and substituting the equations above, we find the critical wage ($\hat{w}$):

$$w \ge rV_u + \frac{(r + b + q)e}{q}$$

**Intuition:** The wage must be high enough that the opportunity cost of being fired $(V_E - V_u)$ outweighs the temporary utility gain of shirking ($e$).

In equilibrium, all firms are identical given competition and pay the same wage. This means we need to find the job acquisition rate ($a$) that keeps the labor market in a "steady state".  

- Flow Balance: The number of people losing jobs ($bL$) must equal the number of people finding jobs ($a(N-L)$).
- Solving for $a$: $$a = \frac{bL}{N-L}$$


Substituting this $a$ back into the worker's NSC equation gives us the Aggregate No-Shirking Condition:

$$w \ge e + \bar{w} + \frac{e}{q} \left( \frac{b}{u} + r \right)$$

where $u$ is the unemployment rate $(N-L)/N$.  

When $u \to 0$ the Aggregate NSC equation is equivalent to full employment. As unemployment disappears, the job acquisition rate $a$ goes to infinity. This means a fired worker is rehired instantly.

- If you are rehired instantly, there is no penalty for being fired.
- If there is no penalty, the only way to stop you from shirking is an infinitely high wage ($w \to \infty$).
- Since firms cannot pay infinite wages, equilibrium must involve some unemployment to make the threat of firing credible.

The **final equilibrium** is the point where the Aggregate Labor Demand curve ($F'(L)$) intersects the Aggregate NSC

1. **Demand:** Firms hire until the marginal product of labor equals the wage.
2. **Supply (NSC):** Workers only provide effort if the wage is above the NSC curve.
3. **Result:** The equilibrium wage $w^*$ is higher than the market-clearing wage, and the equilibrium employment $L^*$ is lower than the total labor supply $N$. This gap ($N - L^*$) is involuntary unemployment.

Main results per parameter:

- If Monitoring Improves ($q \uparrow$): The NSC shifts down; wages fall and employment rises. This is driven by the fact that firms can better monitor the effort that employes put in and therefore can give less benefits (a lower wage)  without people stop putting in effort.
- If Benefits Increase ($\bar{w} \uparrow$): The NSC shifts up; wages rise and unemployment increases because the punishment of being fired is less severe. In other words, cost opportunity of unemployement is lower and firms have to give a higher wage to make sure they put effort. Since firms pay the marginal productivity of labor, less firms are able to pay this higher wage. This leads to a higher level of overall unemployement.
- If Interest Rates Rise ($r \uparrow$): Workers care less about the future (the future loss of being fired), so firms must pay more today to keep them honest.  

## 2. Research question

Throughout the project we want to check different questions. 
The main question is:

- How does the introduction of an informal sector and agent heterogeneity modify the equilibrium dynamics and welfare properties of a labor market governed by efficiency wages?

The secundary questions we try to answer are:

1. How does the presence of an informal sector acting as a fallback option for fired workers influence the No-Shirking Condition (NSC) compared to the standard model where workers only face pure unemployment?
2. In a dynamic economy with heterogeneous firms (varying monitoring intensities $q_i$) and workers (varying discount rates $r_i$), does the labor market converge to the analytical steady-state equilibrium?
3. How would a multi shock (in various parameters) would look like in this model? How would it replicate the case of implementation of AI in the economy?
4. To what extent can the modified efficiency wage model replicate historical labor market dynamics in emerging economies, and what are the implied levels of productivity ($A$) and search efficiency ($\lambda_i$) over time?

## 3. Motivation

The motivation for these research questions lies in bridging the gap between high-level economic theory and the complex realities of modern labor markets. By extending the Shapiro-Stiglitz (1984) framework, we move from a stylized "black and white" model of employment to a more nuanced "grey" reality that includes informality and technological shocks.  

**Main Question: The Dynamics of a Dual-Sector Economy**

Most standard macroeconomic models assume a binary labor market: you are either employed in a firm or receiving state-sponsored benefits. However, in many global economies, specially in the global south, the safety net is not a government check but a vibrant informal sector. Understanding how these two sectors interact is essential for designing policies that don't just fix unemployment but actually improve the quality of jobs available.

**Secondary Question I: The Endogenous Fallback Option**

In the original 1984 model, the threat of firing is constant because the alternative—pure unemployment—is static. Economically, if a worker knows they can immediately start selling goods or services in the informal market, the penalty for being fired changes. This question motivates a study of how endogenous outside options affect the wage floor that formal firms must set to prevent shirking. This might also influence the way policies are made since it can have an ambiguous effect and welfare oriented policies may lead to higher levels of informal employement and lower formal employement.

**Secondary Question II: Agent Heterogeneity and Market Convergence**

Theoretical economics often relies on the "representative agent" assumption—the idea that every worker and firm behaves exactly the same. By using an Agent-Based Model (ABM), we test if the market still reaches a stable equilibrium when people have different levels of patience ($r_i$) or firms have different levels of technology ($q_i$). This provides a "stress test" for the structural integrity of the Shapiro-Stiglitz theory.

**Secondary Question III: The "AI Revolution" Multi-Parameter Shock**

Technology is rarely a single-parameter change. The AI revolution is expected to increase productivity ($A$), but it also radically changes how workers are monitored ($q$) and increases the volatility of job roles ($b$). This question is motivated by the need to understand "competing effects": if AI makes you more productive but also easier to watch, what would happened in a real economy?

**Secondary Question IV: Empirical Calibration (Colombia)**

A model's true value is its ability to explain the world. By calibrating the model to Colombia’s labor data, we can "unmask" the hidden drivers of the economy. If the model shows that informal employment rose while formal employment stayed flat, we can use the math to calculate the implicit productivity (A) that must have existed to cause that result. This transforms a theoretical tool into a diagnostic one.

## 4. Preview of the results

The introduction of an informal sector shifts the equilibrium from a binary "Work vs. Unemployment" state to a "Work vs. Non-Formal Pool" state. The market equilibrium is reached when the Marginal Product of Labor (MPL) intersects the Aggregate No-Shirking Condition (NSC):

$$A \alpha L_f^{\alpha - 1} = e + w_{fallback} + \frac{e}{q} \left( \frac{b}{u_{total}} + r \right)$$

- Economic Result: In the presence of a productive informal sector, $w_{fallback} > \bar{w}$, forcing formal firms to pay higher efficiency wages to maintain the "penalty" for shirking.
- Mathematical Logic: Because the fallback utility is higher, the capital loss $(V_E - V_u)$ associated with being fired is smaller for any given wage, requiring a higher $\hat{w}$ to satisfy the NSC inequality $V_E^N \ge V_E^S$

An interesting hypothesis that surges from this results is that a higher state control against informal employement can actually increase the number of formal workers. This since it can make the rate of finding an informal job lower and therefore, it can make the efficiency wages lower (increasing the number of jobs firms are willing to give). It is important to say that we are not against informal jobs but they have some big problems for economies since in most countries they don't contribute to healthcare or pension systems, stressing them out financially. It is therefore important to find solutions to formalize informal workers and integrate them to the systems (also since it becomes a bigger and less fragile safety net for them).

**Secondary Question Previews**

**I. The Endogenous Fallback Wage ($w_{fallback}$)**:
- Result: The fallback wage is no longer a fixed benefit but a dynamic variable.  Math: $w_{fallback}$ is calculated as a weighted average of benefits and informal earnings:$$w_{fallback} = \frac{U \cdot \bar{w} + L_i \cdot w_{inf}}{U + L_i}$$

- Economic Impact: When the informal sector ($L_i$) grows relative to pure unemployment ($U$), the reservation utility of formal workers increases, which can paradoxically lead to higher formal unemployment as firms reduce labor demand in response to higher required wages.  

**II. Convergence Under Agent Heterogeneity**

- Result: The Agent-Based Model (ABM) successfully converges to the analytical steady state despite micro-level noise.

- Math: The simulation utilizes stochastic Poisson transitions where the probability of a state change in time $dt$ is $\lambda \cdot dt$. Over $T = 8000$ periods, the law of large numbers ensures that the simulated means for $L_f$ and $w^*$ match the root-finding results of the analytical equations.  Economic Impact: This validates that the "Worker Discipline Device" remains a stable market attractor even when individual workers have different discount rates ($r_i$).

**III. The AI Revolution Multi-Shock**

- Result: A simultaneous increase in Productivity ($A$) and Monitoring ($q$) offsets the negative impact of higher Separation Rates ($b$).  

- Math: The shock increases $A$ (shifting labor demand up) and $q$ (shifting the NSC curve down), while the increase in $b$ (shifting the NSC up) is mathematically dominated by the productivity gains in the simulated parameters.  

- Economic Impact: The "AI Revolution" leads to a transition where formal employment ($L_f$) expands, and wages rise, while the informal sector shrinks as workers are pulled into more productive formal roles.

**Empirical Insights (The Colombia Calibration)**

- Result: The model replicates historical trends by treating productivity ($A$) and informal search speed ($\lambda_i$) as unobserved parameters to be optimized.

- Math: The calibration minimizes the Sum of Squared Errors (SSE):$$\min_{A, \lambda_i} \sum_{k \in \{L_f, L_i, U\}} (k_{real} - k_{model})^2$$

- Economic Impact: The results suggest that the Informality Rate in economies like Colombia is an endogenous response to the implicit productivity gap between sectors. ## 5. Main assumptions

This model have several major assumptions:

**Core model main/strong assumptions:**

- Identical Agents: There are a fixed number ($N$) of identical workers and ($M$) identical firms. This ensures that being fired carries no social stigma, as a fired worker is not inherently "worse" than any other. In other words, all workers have the same quality and same education.
- Risk Neutrality: Workers are risk-neutral, with a separable utility function $U = w - e$, where $w$ is the wage and $e$ is the effort level.
- Infinite Life & Discounting: Individuals are infinitely lived and discount future utility at a constant rate $r > 0$.
- Imperfect Monitoring: Firms cannot costlessly observe effort. There is an exogenous probability ($q$) per unit of time that a shirker is caught. This may not hold in different sectors.
- Exogenous Separations: Jobs end for reasons unrelated to shirking (e.g., relocation) at a constant rate ($b$). The rate is not necesarilly fixed in an economy.
- Perfect Information on Jobs: Workers have perfect information about all job opportunities; there is no "search unemployment" in the standard model.
- Wealth Constraints: Workers cannot post performance bonds to guarantee effort, either due to a lack of wealth or the risk of firm opportunism.


**Adapted models main/strong assumptions:**

- Endogenous Informal Sector: Unlike the standard model where the alternative to work is pure unemployment, a "non-formal" pool consisting of both pure unemployed ($U$) and informal workers ($L_i$) exists.
- Informal Congestion (Crowding-Out): The informal sector is subject to diminishing returns. The informal wage ($w_{inf}$) drops linearly as participation increases, modeled by the congestion parameter $\gamma_{inf}$:
  $$w_{inf} = \max(\bar{w}, w_{inf\_max} - \gamma_{inf} L_i)$$

- Heterogeneous Firms: In the simulation, firms are not identical. Each firm ($j$) has a unique monitoring intensity $q_j$ drawn from a normal distribution.
- Heterogeneous Workers: Workers have idiosyncratic discount rates $r_i$, meaning their sensitivity to the future loss of being fired varies.
- Cobb-Douglas Production: labor demand is based on the aggregate production function $Q = A \cdot L^\alpha$, where $\alpha$ is labor elasticity. WE assume that firms don't use capital to produce which definitely don't align with reality.
- Profit-Driven Hiring: In the dynamic simulation, the job acquisition rate ($a$) is not just a steady-state result but an active adjustment. Hiring speed increases if the Marginal Product of Labor (MPL) exceeds the current average NSC wage:
$$a = a_{base} \cdot (1.0 + 3.0 \cdot (MPL - w_{avg}))$$
- Transition Paths: We assume specific paths for workers, such as falling into pure unemployment immediately after a formal separation before being able to "find" informal work at rate $\lambda_i$

## 6. The model and the extension

The core theory posits that unemployment serves as a discipline device because employers cannot costlessly monitor effort. 

1. Worker Asset Equations:
Workers are risk-neutral and maximize the present discounted value of utility with discount rate $r$. Instantaneous utility is $U = w - e$, where $w$ is the wage and $e$ is effort.

Workers choose between providing effort ($e > 0$) or shirking ($e = 0$).  

The fundamental equations are defined as:

- Employed Shirker ($V_E^S$): Receives wage $w$ and faces a total job loss rate of $(b + q)$, where $b$ is exogenous separation and $q$ is the probability of being caught shirking.
  $$rV_E^S = w + (b + q)(V_u - V_E^S)$$

- Employed Non-Shirker ($V_E^N$): Receives wage $w$, exerts effort $e$, and only faces the exogenous separation rate $b$.

  $$rV_E^N = w - e + b(V_u - V_E^N)$$


- Unemployed Individual ($V_u$): Receives benefits $\bar{w}$ and finds a job at rate $a$.

  $$rV_u = \bar{w} + a(V_E - V_u)$$


2. Mathematical Demonstration of the No-Shirking Condition (NSC)

To elicit effort, the firm must ensure $V_E^N \ge V_E^S$. 

We solve for the critical wage $\hat{w}$ where $V_E^N = V_E^S$:  Rearrange the equations to solve for the values: 

$$V_E^S = \frac{w + (b + q)V_u}{r + b + q}$$ 
$$V_E^N = \frac{w - e + bV_u}{r + b}$$  

Set them equal and simplify:

$$\frac{w - e + bV_u}{r + b} = \frac{w + (b + q)V_u}{r + b + q}$$ 
$$(w - e + bV_u)(r + b + q) = (w + (b + q)V_u)(r + b)$$  

The Resulting NSC: Solving for $w$ yields the individual firm's required wage:

$$w \ge rV_u + \frac{(r + b + q)e}{q}$$  

3. Aggregate NSC in Steady StateIn equilibrium, the flow into unemployment ($bL$) must equal the flow out ($a(N-L)$).

Solving for the job acquisition rate gives $a = \frac{bL}{N - L}$. 
Substituting $a$ and $V_u$ into the NSC yields the Aggregate No-Shirking Constraint:
  $$w \ge e + \bar{w} + \frac{e}{q} \left( \frac{b}{u} + r \right)$$  
Where $u$ is the unemployment rate $(N-L)/N$.  

**The Extension:** 

Endogenous Informality: replaces the fixed unemployment benefit $\bar{w}$ with a dynamic fallback wage ($w_{fallback}$) derived from a secondary labor market.  
1. The Endogenous Fallback Logic:

In this version, a worker who loses a formal job does not just face pure unemployment ($U$) but can also enter the informal sector ($L_{inf}$).
The expected utility of a non-formal worker is weighted by the probability of being in either state:

  $$w_{fallback} = \frac{U \cdot \bar{w} + L_{inf} \cdot w_{inf}}{U + L_{inf}}$$  
  
2. Informal Sector Congestion MathThe informal wage ($w_{inf}$) is not fixed; it decreases as the sector becomes crowded (diminishing returns). This is modeled as:

  $$w_{inf} = \max(\bar{w}, w_{inf\_max} - \gamma_{inf} L_{inf})$$

$w_{inf\_max}$: The maximum wage in the informal sector when it is empty. 
$\gamma_{inf}$: The rate at which wages drop due to crowding.  

3. Modified Transition Flows

The steady state now requires balancing flows between three states: Formal ($L_f$), Informal ($L_i$), and Unemployed ($U$).  
Informal flow: Workers find informal work at rate $\lambda_i$ and lose it at rate $\delta_i$.

  $$\lambda_i U = (\delta_i + a) L_i$$  

Job acquisition ($a$): Now, both unemployed and informal workers compete for formal jobs at rate $a$.

  $$a = \frac{b L_f}{U + L_i}$$  

4. The Resulting EquilibriumThe market equilibrium in the extension is found by solving for $L_f$ such that the Marginal Product of Labor (MPL) equals the modified NSC:

  $$A \alpha L_f^{\alpha - 1} = e + w_{fallback} + \frac{e}{q} \left( \frac{b}{u_{total}} + r \right)$$  
  
Where $u_{total} = (N - L_f)/N$.

## Efficiency wages Shapiro - Stiglitz base model 

Here we define the structural parameters of the **Shapiro-Stiglitz (1984) Efficiency Wage Model**. In this framework, workers choose whether to exert effort or "shirk" (work below standard). Firms cannot perfectly observe effort, so they must pay a wage premium to deter shirking.

- `N = 1.0`: Total labor supply (normalized to 1), representing the full workforce available.
- `e = 0.1`: Effort cost — the disutility a worker incurs from working diligently. Higher `e` means workers require a larger wage premium to be willing to work.
- `r = 0.05`: Discount rate — how much workers value future income versus the present. A higher discount rate makes the threat of unemployment less effective, requiring higher wages.
- `b = 0.1`: Exogenous separation rate — the rate at which employed workers lose their jobs regardless of behavior (e.g., due to firm closures or health reasons).
- `q = 0.5`: Monitoring intensity — the probability that a shirking worker is caught. Higher `q` reduces the wage firms must pay to deter shirking.
- `w_bar = 0.0`: Unemployment benefit — the income a worker receives when unemployed. Higher `w_bar` raises the NSC wage.
- `alpha = 0.7`: Labor elasticity in production (from a Cobb-Douglas production function `F(L) = A*L^alpha`).
- `A = 1.0`: Total Factor Productivity (TFP) — a scaling factor for output.

```python
params = {
    'N': 1.0,       # Total labor supply
    'e': 0.1,       # Required effort (disutility)
    'r': 0.05,      # Discount rate
    'b': 0.1,       # Exogenous separation rate
    'q': 0.5,       # Monitoring intensity (prob. of catching a shirker)
    'w_bar': 0.0,   # Unemployment benefit
    'alpha': 0.7,   # Labor elasticity in production
    'A': 1.0        # Total factor productivity
}
```

This function computes the **minimum wage** that firms must pay to prevent workers from shirking, given an employment level `L`. This is the core equation of the Shapiro-Stiglitz model.

A rational worker will exert effort only if the expected value of employment (with effort) exceeds the expected value of shirking. The NSC wage equates the two:

`w* = w_bar + e + (e/q) * (b/u + r)`

Where:
- `u = (N-L)/N` is the **unemployment rate** — the fraction of workers without jobs. Higher unemployment increases the cost of job loss, making the threat of termination more credible, and thereby lowering the required wage premium.
- The term `(e/q) * (b/u + r)` represents the **efficiency wage premium** — the extra wage needed above the unemployment benefit to make effort rational.
- As `u → 0` (full employment), the wage must rise to infinity — workers face no unemployment threat and would always shirk at lower wages. This is why full employment is unattainable in this model.

We make `u` the maximum between u and $1e-6$ to avoid divisions by zero.

```python
def nsc_wage(L, p):

    u = (p['N'] - L) / p['N']
    u = np.maximum(u, 1e-6)
    
    return p['e'] + p['w_bar'] + (p['e'] / p['q']) * (p['b'] / u + p['r'])
```

This function computes the **wage that profit-maximizing firms are willing to pay** for a given level of employment `L`. Under perfect competition, firms hire until the wage equals the marginal product of labor (MPL).

**Economic intuition:** Assuming a Cobb-Douglas production function `F(L) = A * L^alpha`, the MPL is:

`MPL = F'(L) = A * alpha * L^(alpha-1)`

This is a **downward-sloping** function of `L`: as firms hire more workers, the marginal contribution of each additional worker decreases (diminishing returns to labor). The labor demand curve slopes downward in wage-employment space.

```python
def labor_demand_mpl(L, p):
    return p['A'] * p['alpha'] * L**(p['alpha'] - 1)
```

This function numerically solves for the **Shapiro-Stiglitz equilibrium** — the point where the NSC wage equals the firm's labor demand (MPL). This intersection determines the equilibrium wage `W*` and employment level `L*`.

**Economic intuition:** In equilibrium:
- Firms are willing to pay `W*` for `L*` workers (their MPL equals `W*`).
- Workers at wage `W*` are just indifferent between shirking and exerting effort (the NSC binds).
- The equilibrium features **involuntary unemployment**: `L* < N`, meaning some workers want to work at the going wage but cannot find jobs. This unemployment is what disciplines workers and makes the efficiency wage sustainable.

The solver tries multiple initial guesses to ensure robustness and avoid missing the equilibrium.

```python
def find_equilibrium(p):
    obj = lambda L: labor_demand_mpl(L, p) - nsc_wage(L, p)
    try:
        for x0 in [0.5, 0.3, 0.7, 0.1, 0.9]:
            L_eq = fsolve(obj, x0, full_output=True)
            if L_eq[2] == 1:
                L_sol = float(L_eq[0][0])
                if 0 < L_sol < p['N']:  
                    W_sol = nsc_wage(L_sol, p)
                    return L_sol, W_sol
        return np.nan, np.nan
    except:
        return np.nan, np.nan
```

Here we compute the values of both the **NSC wage** and the **Marginal Product of Labor** across a fine grid of employment levels from near-zero to near-full employment.

**Economic intuition:** This prepares the data to plot the two central curves of the Shapiro-Stiglitz diagram:
- The **NSC curve** (`nsc_vals`) is upward-sloping in employment: as more workers are employed, unemployment falls, reducing the penalty of job loss, so firms must pay more to deter shirking.
- The **MPL curve** (`mpL_vals`) is downward-sloping: as employment rises, each additional worker contributes less to output (diminishing marginal product).

The intersection of these two curves gives the equilibrium.

```python
L_range = np.linspace(0.01, 0.999, 300)
nsc_vals = nsc_wage(L_range, params)
mpL_vals = labor_demand_mpl(L_range, params)
```

We now solve for the equilibrium values by calling the `find_equilibrium` function.

**Economic intuition:** The result `(L_eq, W_eq)` is the unique pair of employment and wage where:
1. Firms are maximizing profits (wage = MPL)
2. Workers are just indifferent between shirking and working (NSC holds with equality)

This equilibrium wage `W_eq` is **above the market-clearing wage**, creating involuntary unemployment of `N - L_eq` workers. This is the key prediction of efficiency wage theory: firms willingly pay above-market wages to discipline the workforce.

```python
L_eq, W_eq = find_equilibrium(params)
```

This cell generates the **canonical Shapiro-Stiglitz diagram** showing the equilibrium graphically.

The plot displays:
- **NSC curve** (upward-sloping): the minimum wage required at each employment level to prevent shirking. It rises steeply as employment approaches full capacity, reflecting the vanishing threat of unemployment.
- **Labor Demand / MPL curve** (downward-sloping): the maximum wage firms are willing to pay, determined by workers' marginal productivity. Firms always hire where wage = MPL.
- **Equilibrium point ($L^*$, $W^*$):** the intersection where both conditions hold simultaneously. The equilibrium wage exceeds what the market would generate without information asymmetry, and unemployment is an equilibrium phenomenon, not a market failure.

```python
plt.figure(figsize=(9, 6))
plt.plot(L_range, nsc_vals, color='steelblue', linewidth=2.5, label='NSC: No-Shirking Wage')
plt.plot(L_range, mpL_vals, color='darkorange', linewidth=2.5, label='Labor Demand (MPL)')

if not np.isnan(L_eq):
    plt.scatter([L_eq], [W_eq], color='red', s=120, zorder=5, label=f'Equilibrium E: L*={L_eq:.3f}, w*={W_eq:.3f}')
    plt.axvline(L_eq, color='red', linestyle=':', alpha=0.5)
    plt.axhline(W_eq, color='red', linestyle=':', alpha=0.5)
    plt.axvline(params['N'], color='gray', linestyle='--', alpha=0.5, label='Full Employment (N=1)')

plt.xlabel('Employment (L)', fontsize=12)
plt.ylabel('Wage (w)', fontsize=12)
plt.title('Shapiro-Stiglitz Model: NSC Curve and Labor Demand (Figure 2)', fontsize=13)
plt.legend(fontsize=10)
plt.ylim(0, min(2.0, np.percentile(nsc_vals, 95) * 1.5))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

We now conduct a comparative statics exercise by varying `q` — the **monitoring intensity** (probability of catching a shirker).

The monitoring parameter `q` directly affects the effectiveness of the unemployment threat. When `q` is higher:
- Firms are more likely to detect and fire shirkers.
- The expected payoff from shirking falls, reducing the wage premium needed to deter it.
- The NSC curve **shifts downward**, leading to higher equilibrium employment and lower equilibrium wages.

This illustrates a key policy implication: investing in monitoring technology can reduce structural unemployment in efficiency wage economies.

```python
q_range = np.linspace(0.1, 0.9, 50)
L_q, W_q = [], []
for val in q_range:
    p_test = params.copy()
    p_test['q'] = val
    L_eq, W_eq = find_equilibrium(p_test)
    L_q.append(L_eq)
    W_q.append(W_eq)
```

This subplot shows how the **equilibrium wage and employment** change as `q` (monitoring probability) varies.

The comparative statics reveal:
- **Right panel (Equilibrium Wage vs. q):** As monitoring improves, firms need to pay less to deter shirking — wages fall. This is because better monitoring substitutes for the wage premium as a disciplining device.
- **Left panel (Equilibrium Employment vs. q):** Lower wages mean firms hire more workers, so employment rises toward the full-employment benchmark.

This demonstrates the **fundamental trade-off** in efficiency wage models: the same unemployment that disciplines workers also constitutes a social inefficiency.

This function computes the **total output** of the economy given employment `L` and a parameter set `p`.

Output is produced according to a Cobb-Douglas function:

`F(L) = A * L^alpha`

This captures two key economic facts:
- **Increasing in L**: More workers produce more output.
- **Diminishing returns**: Each additional worker adds less to output than the previous one (since `alpha < 1`), which drives the downward-sloping labor demand curve.

```python
def production_total(L, p):
    return p['A'] * L**p['alpha']
```

This function computes **social welfare** — total output minus total labor costs (wages paid).

Social welfare here captures the **net surplus** generated by the economy:

`Welfare = F(L) - W * L = A * L^alpha - w * L`

This is the difference between what the economy produces and what it costs in wages. In the efficiency wage equilibrium, this surplus is reduced because:
1. The wage is above the competitive level (efficiency wage premium).
2. Employment is below the socially optimal level (structural unemployment).

The social planner would choose higher employment and lower wages, but this is not incentive-compatible without monitoring.

```python
def social_welfare(L, p):
    return production_total(L, p) - p['e'] * L
```

This function computes the **socially optimal employment level** — the level that maximizes total welfare without the constraint of the NSC.

The social planner maximizes `F(L) - cost_per_worker * L`. Without the moral hazard problem (shirking), the planner could achieve full employment at a lower wage. The difference between the social optimum and the efficiency wage equilibrium quantifies the **welfare cost of moral hazard** — the loss in output and employment due to imperfect monitoring.

```python
def find_social_optimum(p):
    def neg_welfare(L):
        if L <= 0 or L >= p['N']:
            return 0
        # Check NSC with wage = APL
        w_apl = production_total(L, p) / L
        w_nsc = nsc_wage(L, p)
        if w_apl < w_nsc:
            return 0  # NSC violated
        return -social_welfare(L, p)
    
    result = minimize_scalar(neg_welfare, bounds=(0.01, p['N']-0.001), method='bounded')
    L_s = result.x
    return L_s, social_welfare(L_s, p)
```

We plot how **equilibrium welfare** changes as monitoring intensity `q` varies.

Better monitoring:
1. Reduces the required efficiency wage (firms pay less per worker)
2. Increases equilibrium employment (more workers are hired)
3. Increases total output (more production)
4. Decreases total wage bill (lower wages, but to more workers)

The net effect on welfare is generally positive: better monitoring moves the economy closer to the social optimum.

```python
q_range_welfare = np.linspace(0.2, 0.9, 50)
welfare_market = []
welfare_optimal = []

for val in q_range_welfare:
    p_test = params.copy()
    p_test['q'] = val
    L_m, _ = find_equilibrium(p_test)
    w_m = social_welfare(L_m, p_test) if not np.isnan(L_m) else np.nan
    welfare_market.append(w_m)
    L_s, w_s = find_social_optimum(p_test)
    welfare_optimal.append(w_s)
```

This visualization overlays the **efficiency wage market equilibrium** and the **social planner's optimum** on the Shapiro-Stiglitz diagram.

The gap between the two solutions reveals the **deadweight loss** from moral hazard:
- The market equilibrium has too-high wages and too-low employment.
- The social optimum achieves higher employment at lower wages, but is unattainable without perfect monitoring (or first-best contracts).
- This gap motivates policy interventions: unemployment insurance reform, monitoring subsidies, or minimum effort standards.

```python
plt.figure(figsize=(9, 6))
plt.plot(q_range_welfare, welfare_market, label='Market Welfare', color='red', linewidth=2)
plt.plot(q_range_welfare, welfare_optimal, label='Social Optimum Welfare', color='green',
         linestyle='--', linewidth=2)
plt.fill_between(q_range_welfare, welfare_market, welfare_optimal,
                 color='gray', alpha=0.2, label='Efficiency Loss (DWL)')
plt.xlabel('Monitoring Intensity (q)', fontsize=12)
plt.ylabel('Social Welfare = F(L) - eL', fontsize=12)
plt.title('Social Welfare: Market vs. Social Planner Optimum', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Stiglitz Base - Simulated Dynamic Economy

We define the number of workers and firms in our **agent-based simulation** of the efficiency wage economy.

Rather than solving for the aggregate equilibrium analytically, we now simulate an economy with discrete agents:
- `N_workers`: the labor force — each agent represents a worker who decides each period whether to exert effort or shirk.
- `N_firms`: the number of employers — each firm posts wages and hires/fires workers based on monitoring outcomes.

This approach helps us see how the macro equilibrium (NSC, unemployment, wage premium) emerges from individual behavior rather than being imposed top-down.

```python
N_workers = 10000      
N_firms = 1000         
T = 600              
dt = 0.1             
```

Here we define the parameters for the **dynamic agent-based simulation**.

**Economic intuition:** These parameters govern the temporal dynamics of the efficiency wage economy:
- `T` (time periods): how many rounds the simulation runs — each period, workers decide to work or shirk, firms monitor, and hiring/firing occurs.
- `w_firm` (firm wage): the wage each firm posts. In a competitive equilibrium, this converges to the efficiency wage `W*`.
- `theta` (job finding rate): the speed at which unemployed workers find new jobs — this affects the expected duration of unemployment and thus the cost of job loss.

This dynamic setup captures the **labor market flows** that the static Shapiro-Stiglitz model abstracts away.

- `wages[t]`: the wage paid in each period.

Tracking these histories lets us compute aggregate statistics (unemployment rate, wage levels, output) and observe how the efficiency wage equilibrium is maintained dynamically.

4. Unemployed workers find new jobs at rate `theta`.

This mirrors the continuous-time Shapiro-Stiglitz model in discrete time. The simulation demonstrates that the NSC wage is not just a theoretical construct — it emerges naturally when rational agents interact repeatedly in a labor market with imperfect monitoring. The resulting steady-state unemployment and wage levels should closely match the analytical solution.

## Stiglitz with informal sector

We define the parameters of the Stiglitz model incorporating the informal sector. 

In economic terms:
- **`N=1.0`**: labour force normalised to 1 (as a proportion of the population).
- **`e, r, b, q`**: parameters of the shirking problem: required effort, discount rate, formal separation rate and probability of monitoring.
- **`w_bar`**: unemployment benefit (reservation wage in the informal sector/unemployment).
- **`alpha, A`**: parameters of the Cobb-Douglas production function `Y = A * L^alpha`.
- **`w_inf_max, gamma_inf`**: maximum wage in the informal sector and the rate at which it falls due to crowding-out.
- **`lambda_i, delta_i`**: transition rates to and from informal employment (Poisson process).

```python
params = {
    'N': 1.0,           
    'e': 0.1,           
    'r': 0.05,          
    'b': 0.15,          
    'q': 0.5,           
    'w_bar': 0.05,      
    'alpha': 0.9,       
    'A': 0.4,           
    
    'w_inf_max': 0.35,  
    'gamma_inf': 1.2,   
    'lambda_i': 0.12,   
    'delta_i': 0.05     
}
```

`nsc_wage(L_formal, p, w_fallback)` implements the updated NSC from the Shapiro-Stiglitz model with an informal sector. The formal minimum wage that prevents shirking is:

`w_NSC = e + w_fallback + (e/q) * (b/u_total + r)`

Where `w_fallback` dynamically replaces the unemployment benefit `w_bar` with the worker’s expected wage if they lose their formal job, weighting the probability of falling into informality versus pure unemployment. This is key: the informal sector compresses the wage premium by increasing the worker’s opportunity cost.

```python
def nsc_wage(L_formal, p, w_fallback):

    u_total = (p['N'] - L_formal) / p['N']          
    u_total = np.maximum(u_total, 1e-6)            
    
    return p['e'] + w_fallback + (p['e'] / p['q']) * (p['b'] / u_total + p['r'])
```

`labor_demand_mpl(L_formal, p)` calculates the Marginal Product of Labour (MPL) for formal firms under Cobb-Douglas technology:

`MPL = A * alpha * L^(alpha - 1)`

At equilibrium, the firm hires until `MPL = w_NSC`. The function returns the wage that a representative firm is willing to pay given the employment level `L_formal`. 

Same as before, `max(L_formal, 1e-6)` prevents division by zero at the extremes.

```python
def labor_demand_mpl(L_formal, p):
    L_formal = max(L_formal, 1e-6) 
    return p['A'] * p['alpha'] * L_formal**(p['alpha'] - 1)
```

`get_ss_Li(L_f, p)` calculates the long-term level of informality `L_i^*` consistent with a level of formal employment `L_f`. It is based on the flow balance of the informal market (Beveridge equation):

`lambda_i * U = (delta_i + a) * L_i`

Where `a = b * L_f / NF` is the rate of absorption of informal workers into the formal sector (proportional to formal separations). This equation implies that informality is endogenous: more formal employment generates greater pressure to exit the informal sector.

```python
def get_ss_Li(L_f, p):
    """Calculate steady-state informal employment given formal employment."""
    NF = p['N'] - L_f 
    if NF <= 0: return 0

    a = (p['b'] * L_f) / NF
    Li = (p['lambda_i'] * NF) / (p['lambda_i'] + p['delta_i'] + a)
    return Li
```

`analytical_obj(L_f, p)` constructs the residual of the labour market equilibrium condition:

`f(L_f) = NSC(L_f, w_fallback) - MPL(L_f) = 0`

Given `L_f`, it endogenously calculates `L_i`, `U`, `w_inf`, `w_fallback` and then assesses whether the wage that maximises the firm’s utility (PMgL) equals the non-shirking minimum wage (NSC). The zero-crossing of this function defines the market equilibrium.

```python
def analytical_obj(L_f, p):
    Li = get_ss_Li(L_f, p)
    U = p['N'] - L_f - Li
    NF = U + Li
    
    w_inf = max(p['w_bar'], p['w_inf_max'] - p['gamma_inf'] * Li)
    w_fallback = (U * p['w_bar'] + Li * w_inf) / NF if NF > 0 else p['w_bar']
    
    w_nsc = nsc_wage(L_f, p, w_fallback)
    w_mpl = labor_demand_mpl(L_f, p)
    return w_nsc - w_mpl
```

`find_analytical_eq(p)` uses `root_scalar` method to find the `L_f*` that equilibrates the labour market. 
It returns the complete equilibrium vector: `{L_f, L_i, U, w_f, w_inf, w_fallback}`. This equilibrium is the theoretical benchmark against which the agent-based simulation is compared.

```python
def find_analytical_eq(p):
    try:
        res = root_scalar(analytical_obj, args=(p,), bracket=[0.001, p['N']-0.001])
        if res.converged:
            L_f = res.root
            Li = get_ss_Li(L_f, p)
            U = p['N'] - L_f - Li
            w_inf = max(p['w_bar'], p['w_inf_max'] - p['gamma_inf'] * Li)
            w_fallback = (U * p['w_bar'] + Li * w_inf) / (U + Li)
            w_nsc = nsc_wage(L_f, p, w_fallback)
            return {'L_f': L_f, 'L_i': Li, 'U': U, 'w_f': w_nsc, 'w_inf': w_inf, 'w_fallback': w_fallback}
    except ValueError:
        pass
    return None
```

In economic terms, these are the long-run values towards which the model converges if the parameters remain unchanged: the proportion of workers in each sector and the equilibrium wage levels.

```python
eq = find_analytical_eq(params)
```

## Welfare analysis with informality

We calculate total social surplus in the dual-sector economy.
    

Welfare is the total value of everything produced minus the total cost of effort.
    
1. **Formal Output ($Y_f$):** Standard Cobb-Douglas production ($A \cdot L_f^\alpha$).
2. **Informal Output ($Y_i$):** Because the informal wage drops linearly as more people enter ($w_{inf} = w_{inf\_max} - \gamma_{inf} \cdot L_i$), the total output of this sector is the area under this marginal productivity curve (an integral resulting in a quadratic function).
3. **Unemployment Benefit ($U \cdot \bar{w}$):** The baseline value/leisure enjoyed by the purely unemployed.
4. **Effort Cost ($e \cdot L_f$):** Only formal workers exert the modeled effort $e$, which is subtracted.

```python
def calculate_social_welfare(L_f, L_i, params):
    """
    Calculates total social surplus in the dual-sector economy.
    W = Formal Output + Informal Output - Formal Effort + Unemployed Benefit
    """
    Y_f = params['A'] * (L_f ** params['alpha'])
    Y_i = params['w_inf_max'] * L_i - 0.5 * params['gamma_inf'] * (L_i**2)
    
    U = max(0.0, 1.0 - L_f - L_i)
    
    return Y_f + Y_i - (params['e'] * L_f) + (U * params['w_bar'])
```

Now we calculate welfare in the standard Shapiro-Stiglitz model (No informal sector).
    
This serves as our "Baseline". If the informal sector didn't exist, anyone who isn't formally employed produces nothing and just receives the baseline unemployment benefit (w_bar). Comparing this to `calculate_social_welfare` reveals the true value of the informal market.

```python
def calculate_standard_welfare(L_f, params):
    """
    Calculates welfare in the standard Shapiro-Stiglitz model (No informal sector).
    """
    Y_f = params['A'] * (L_f ** params['alpha'])
    U = max(0.0, 1.0 - L_f)
    return Y_f - (params['e'] * L_f) + (U * params['w_bar'])
```

Finds the Social Planner's optimal labor allocation.
    
A theoretical "Social Planner" can perfectly observe effort and doesn't need to pay efficiency wages to prevent shirking. They simply allocate workers to maximize total output. We minimize to find the exact point where the marginal product of the formal sector equals the marginal product of the informal sector.

we look for the Formal Employment (L_f) for the standard Shapiro-Stiglitz model.

Finds the root where Labor Demand (MPL) exactly intersects the No-Shirking Condition (NSC). Since there is no informal sector here, the penalty for shirking is falling back into pure unemployment (w_bar).

We look for the exact point where the marginal product of the formal sector equals the marginal product of the informal sector.

```python
def find_first_best_allocation(params):
    """
    Finds the Social Planner's optimal labor allocation (No Information Asymmetry).
    """
    objective = lambda x: -calculate_social_welfare(x[0], x[1], params)
    bounds = ((0.001, 0.999), (0.001, 0.999))
    constraint = {'type': 'ineq', 'fun': lambda x: 1.0 - x[0] - x[1]}
    
    result = minimize(objective, [0.4, 0.4], bounds=bounds, constraints=[constraint])
    
    if result.success:
        return result.x[0], result.x[1], -result.fun
    else:
        raise ValueError("Optimization failed to find First-Best allocation.")
```

```python
base_params = {
    'A': 1.0, 'alpha': 0.7, 'e': 0.1, 'w_bar': 0.05, 
    'w_inf_max': 1.2, 'gamma_inf': 1.2, 'q': 0.5, 'b': 0.1, 'r': 0.05
}
```

We get the formal employment (L_f) for the standard Shapiro-Stiglitz model.
    
    
Finds the point where Labor Demand (MPL) exactly intersects the No-Shirking Condition (NSC). Since there is no informal sector here, the penalty for shirking is falling back into pure unemployment (w_bar).

```python
def get_standard_equilibrium(params):
    """Finds L_f for the standard Shapiro-Stiglitz model (intersection of MPL and NSC)."""
    def nsc_gap(L):
        mpl = params['A'] * params['alpha'] * (L**(params['alpha']-1))
        nsc = params['w_bar'] + params['e'] + (params['e']/params['q']) * (params['b']/(1-L) + params['r'])
        return mpl - nsc
    
    res = root_scalar(nsc_gap, bracket=[0.01, 0.99], method='brentq')
    return res.root if res.converged else 0.01
```

Now we get the Dual-Sector Equilibrium by linking the formal NSC to the informal market wage.
    
1. Informal Absorption: Workers enter the informal sector until the wage drops (due to congestion) to the level of pure unemployment (w_bar).
2. Expected Fallback: If a formal worker is fired, their outside option is no longer just w_bar. It is a weighted average of finding an informal job vs staying purely unemployed.
3. Labor Demand: Formal firms still hire until Wage = MPL.
4. The New NSC: Because the worker's fallback option is now much better, formal firms MUST pay a higher efficiency wage to keep them scared of being fired. 

Result: This higher wage forces formal firms to hire fewer people (L_f drops).

```python
def get_decentralized_equilibrium(params):
    """
    Finds the Dual-Sector Equilibrium by linking the formal NSC to the informal market wage.
    This properly models the "Crowding Out" (discipline) effect.
    """
    def equilibrium_gap(L_f):
        L_i_target = (params['w_inf_max'] - params['w_bar']) / params['gamma_inf']
        L_i = max(0.0, min(L_i_target, 1.0 - L_f - 0.001))
        
        w_inf = params['w_inf_max'] - params['gamma_inf'] * L_i
        w_fallback = max(params['w_bar'], w_inf)
        
        mpl = params['A'] * params['alpha'] * (L_f**(params['alpha']-1))
        
        nsc = w_fallback + params['e'] + (params['e']/params['q']) * (params['b']/(1-L_f) + params['r'])
        
        return mpl - nsc

    res = root_scalar(equilibrium_gap, bracket=[0.01, 0.99], method='brentq')
    
    if res.converged:
        L_f_eq = res.root
        L_i_target = (params['w_inf_max'] - params['w_bar']) / params['gamma_inf']
        L_i_eq = max(0.0, min(L_i_target, 1.0 - L_f_eq - 0.001))
        return L_f_eq, L_i_eq
    else:
        return 0.01, 0.0
```

Now we check for the two main parameters that can change for informal employement. The idea is to check ceteris paribus the formal sector what would happend to the welfare given changes in the informal employement.

```python
gamma_range = np.linspace(0.5, 3.0, 50)
welfare_dual_gamma = []
welfare_std_gamma = []

for gamma in gamma_range:
    current_params = base_params.copy()
    current_params['gamma_inf'] = gamma
    
    L_f_dual, L_i_dual = get_decentralized_equilibrium(current_params)
    welfare_dual_gamma.append(calculate_social_welfare(L_f_dual, L_i_dual, current_params))
    
    L_f_std = get_standard_equilibrium(current_params)
    welfare_std_gamma.append(calculate_standard_welfare(L_f_std, current_params))
```

```python
winf_range = np.linspace(0.1, 1.8, 50)
welfare_dual_winf = []
welfare_std_winf = []

for winf in winf_range:
    current_params = base_params.copy()
    current_params['w_inf_max'] = winf
    
    L_f_dual, L_i_dual = get_decentralized_equilibrium(current_params)
    welfare_dual_winf.append(calculate_social_welfare(L_f_dual, L_i_dual, current_params))
    
    L_f_std = get_standard_equilibrium(current_params)
    welfare_std_winf.append(calculate_standard_welfare(L_f_std, current_params))
```

Based on the resulting graphs, the economy will always be better off when there is an informal market than when there is not. The reason the Informal Premium is entirely green (meaning the dual-sector economy always yields higher or equal welfare than the standard model, with no red crowding-out penalty) stems directly from an assumption in our social welfare function: we assumed that informal work requires zero effort.

Mathematically, our welfare equation is defined as $W = Y_f + Y_i - e \cdot L_f + U \cdot \bar{w}$. Notice that the disutility of effort ($e$) is strictly subtracted for formal workers ($L_f$), but there is no equivalent $- e_{inf} \cdot L_i$ term for the informal sector. 

Economically, when the informal sector becomes highly attractive (e.g., high $w_{inf\_max}$), it forces formal firms to raise wages and shed workers. While this destroys highly productive formal output ($Y_f \downarrow$), society mathematically "saves" the effort cost ($e$) for every worker fired. The displaced workers, alongside the previously unemployed, enter the informal sector where they generate new output ($Y_i$) for "free" (zero effort penalty). Because our modeling decision treats informal productivity as completely costless to the worker's utility, the social surplus gained from putting idle hands to work always mathematically overwhelms the deadweight loss of a shrinking formal sector. To ever see a negative welfare effect (a red zone), we would need to introduce a disutility cost for informal labor, reflecting the real-world physical or precarious toll of working in the shadow economy.

Other improvements and important comments are as follows:

**1. The Missing Tax and Fiscal Wedge**

In the real world, the formal sector pays taxes ($\tau$) that fund public goods and the unemployment benefit ($\bar{w}$). The informal sector might evade these taxes.

We assume $\bar{w}$ is exogenously funded (money falls from the sky).

If workers shift from the formal to the informal sector, the tax base shrinks. To maintain the same $\bar{w}$ for the remaining unemployed, the government must raise taxes on the shrinking formal sector (or on consumption). Our welfare function completely ignores this.

**2. The Absence of Capital ($K$)**

Our formal production function is $Y_f = A \cdot L_f^\alpha$.

What your model does: It assumes labor is the only input. If a worker leaves the formal sector, output drops only by their marginal labor product.

Formal firms rely heavily on Capital ($K$)—machinery, software, infrastructure. If formal employment shrinks because of high efficiency wages, capital sits idle. The lost productivity of idle capital is a catastrophic hit to social welfare that is invisible in a labor-only Cobb-Douglas function.

**3. Linear Congestion vs. Diminishing Returns**

We modeled the informal wage as dropping linearly: $w_{inf} = w_{inf\_max} - \gamma_{inf} L_i$. A linear drop allows the sector to absorb massive amounts of labor while still producing a strictly positive area under the curve (the integral $w_{inf\_max} L_i - 0.5 \gamma_{inf} L_i^2$).

This is not necesarilly true since informal markets (like street vending or gig work) suffer from extreme, non-linear diminishing returns. Once the street corner has 5 vendors, the 6th vendor adds almost zero value to the economy.

**How a Better Model Should Approach This**

**1. Introduce Informal Disutility and Risk ($e_{inf}$ and $\rho$)**

Informal work is often precarious, dangerous, and lacks legal protections. The social welfare function could be modified to include:

- Informal Effort ($e_{inf}$): Often higher than formal effort because informal workers lack labor-saving capital.

- Risk Penalty ($\rho$): A utility discount representing the lack of healthcare, pensions, and legal security.


**2. Endogenize the Government Budget Constraint**

The unemployment benefit $\bar{w}$ must be funded by a payroll tax $\tau$ on formal firms.

Budget Constraint: $\tau \cdot w_f \cdot L_f = \bar{w} \cdot U$

Mechanism: As $L_i$ grows, $L_f$ shrinks. $\tau$ must rise to balance the equation. This increases the cost to formal firms, shifting the labor demand curve (MPL) downward, creating a massive societal penalty for informality.

**3. Implement a Capital-Augmented Production Function**

Use a standard Cobb-Douglas with fixed capital in the short run:

New Production: $Y_f = A \cdot K^{1-\alpha} \cdot L_f^\alpha$

Mechanism: When the informal sector drives up the No-Shirking wage and $L_f$ drops, the marginal product of capital plunges. Society's total output $Y$ will drop much faster.

**4. Convex Informal Congestion**

Include asymptotic decay function that aligns with true physical congestion in shadow economies.

Better Formulation: $w_{inf} = \bar{w} + \frac{w_{inf\_max} - \bar{w}}{1 + \gamma_{inf} \cdot L_i^2}$

This ensures that the marginal addition of informal output rapidly approaches zero, heavily penalizing models that try to shift too much of the workforce out of the formal sector.

We run part of the **dynamic agent-based simulation** of the efficiency wage economy.

The simulation tracks how the economy evolves period by period — how workers respond to wages, how firms react to shirking, and how the labor market reaches (or fluctuates around) the steady-state Shapiro-Stiglitz equilibrium. Each step updates the state variables (employment, wages, output) based on individual agent decisions and random events.

```python
N_workers = 5000      
N_firms = 1000         
T = 8000              
dt = 0.1              
```

```python
q_firms = np.clip(np.random.normal(loc=params['q'], scale=0.05, size=N_firms), 0.05, 0.95)
```

`worker_state` is the central array of the ABM: for each worker `i`, it stores an integer encoding their status in the labour market:
- `0`: fully unemployed
- `-1`: informal worker
- `j > 0` (j = 1..N_firms): formal worker at firm `j`

The initial distribution (60% formal, 20% informal, 20% unemployed) reflects an economy close to the analytically calculated long-run equilibrium.

```python
random_draws = np.random.random(N_workers)
worker_state = np.zeros(N_workers, dtype=int)
worker_state[random_draws < 0.60] = np.random.randint(1, N_firms + 1, np.sum(random_draws < 0.60))
worker_state[(random_draws >= 0.60) & (random_draws < 0.80)] = -1
```

The `history` dictionary is created to record, period by period, the macroeconomic aggregates: the proportions of formal employment (`L_f`), informal employment (`L_i`), unemployment (`U`), and wage levels. These vectors are then used to plot the convergence to equilibrium and compare the results with the analytical values.

```python
history = {
    'formal': [],
    'informal': [],
    'unemployed': [],
    'formal_wage': [],
    'informal_wage': [],
    'fallback_wage': []
}
```

This is the core of the ABM. In each period `t`:
1. The current aggregates are calculated using Boolean masks on `worker_state`.
2. The endogenous informal wage `w_inf(L_i)` (crowding-down) and the fallback wage `w_fallback` are computed as a weighted average.
3. The NSC wage for each firm is determined and states are updated via stochastic Poisson transitions: formal separation (rate `b*dt`), informal separation (rate `delta_i*dt`), formal hiring (rate `a*dt`) and flow into informality (rate `lambda_i*dt`).

This process reflects the microeconomics of the labour market with job search (simplified search-and-matching).

```python
for t in range(T):
    formal_mask = worker_state > 0
    informal_mask = worker_state == -1
    unemp_mask = worker_state == 0
    
    L_f = np.mean(formal_mask)
    L_i = np.mean(informal_mask)
    u_pure = np.mean(unemp_mask)
    non_formal_total = L_i + u_pure
    
    current_w_inf = max(params['w_bar'], params['w_inf_max'] - params['gamma_inf'] * L_i)
    
    if non_formal_total > 0:
        w_fallback = (u_pure * params['w_bar'] + L_i * current_w_inf) / non_formal_total
    else:
        w_fallback = params['w_bar']
        
    j_employed = worker_state[formal_mask] - 1
    if len(j_employed) > 0:
        wages_i = np.array([nsc_wage(L_f, {**params, 'q': q_firms[j]}, w_fallback) for j in j_employed])
        avg_formal_wage = wages_i.mean()
    else:
        avg_formal_wage = np.nan
        
    history['formal'].append(L_f)
    history['informal'].append(L_i)
    history['unemployed'].append(u_pure)
    history['formal_wage'].append(avg_formal_wage if not np.isnan(avg_formal_wage) else 0)
    history['informal_wage'].append(current_w_inf)
    history['fallback_wage'].append(w_fallback)
    
    new_state = worker_state.copy()
    
    emp_idx = np.where(formal_mask)[0]
    if len(emp_idx) > 0:
        sep = np.random.random(len(emp_idx)) < (params['b'] * dt)
        new_state[emp_idx[sep]] = 0  # Fall into pure unemployment initially
        
    inf_idx = np.where(informal_mask)[0]
    if len(inf_idx) > 0:
        inf_sep = np.random.random(len(inf_idx)) < (params['delta_i'] * dt)
        new_state[inf_idx[inf_sep]] = 0
        
    u_idx = np.where(unemp_mask)[0]
    if len(u_idx) > 0:
        gets_inf = np.random.random(len(u_idx)) < (params['lambda_i'] * dt)
        new_state[u_idx[gets_inf]] = -1
    
    avail_idx = np.where(~formal_mask)[0] 
    if len(avail_idx) > 0 and non_formal_total > 0:
        current_mpl = labor_demand_mpl(L_f, params)
        wage_gap = current_mpl - avg_formal_wage if not np.isnan(avg_formal_wage) else 0
        
        a_base = params['b'] * L_f / max(non_formal_total, 0.001)  
        
        a = a_base * (1.0 + 3.0 * wage_gap) 
        a = max(0.0, a)  
        
        a_dt = min(a * dt, 0.99)
        gets_formal = np.random.random(len(avail_idx)) < a_dt
        hired = avail_idx[gets_formal]
        
        if len(hired) > 0:
            new_state[hired] = np.random.randint(1, N_firms + 1, len(hired))
            
    worker_state = new_state
```

The rapid convergence towards these lines confirms the consistency between the agent-based model and the analytical solution: the ABM is a stochastic discretisation of the same equilibrium.

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

time_axis = range(T)
ax1.plot(time_axis, history['formal'], color='steelblue', linewidth=2, label='Formal Employment ($L_f$)')
ax1.plot(time_axis, history['informal'], color='tab:green', linewidth=2, label='Informal Employment ($L_{inf}$)')
ax1.plot(time_axis, history['unemployed'], color='firebrick', linewidth=2, label='Pure Unemployment ($U$)')


ax1.fill_between(time_axis, 0, history['formal'], color='steelblue', alpha=0.1)
ax1.fill_between(time_axis, history['formal'], np.array(history['formal']) + np.array(history['informal']), color='tab:green', alpha=0.1)


if eq:
    ax1.axhline(eq['L_f'], color='steelblue', linestyle=':', linewidth=2, alpha=0.8)
    ax1.axhline(eq['L_i'], color='tab:green', linestyle=':', linewidth=2, alpha=0.8)
    ax1.axhline(eq['U'], color='firebrick', linestyle=':', linewidth=2, alpha=0.8)
    
    ax1.plot([], [], color='gray', linestyle=':', linewidth=2, label='Analytical Equilibrium')

ax1.set_xlabel('Simulation Period', fontsize=11)
ax1.set_ylabel('Population Fraction', fontsize=11)
ax1.set_title('Labor Market Dynamics with Endogenous Informal Sector', fontsize=13)
ax1.legend(fontsize=10, loc='upper right')
ax1.grid(True, alpha=0.3)
ax1.set_ylim(0, 1)


ax2.plot(time_axis, history['formal_wage'], color='steelblue', linewidth=2, label='Avg Formal Wage ($w^*$ NSC)')
ax2.plot(time_axis, history['informal_wage'], color='tab:green', linewidth=2, label='Informal Wage ($w_{inf}$)')
ax2.plot(time_axis, history['fallback_wage'], color='purple', linestyle='--', linewidth=2, label='Expected Fallback Wage ($w_{fallback}$)')
ax2.axhline(params['w_bar'], color='firebrick', linestyle=':', linewidth=2, label='Unemployment Benefit ($w_{bar}$)')


if eq:
    ax2.axhline(eq['w_f'], color='steelblue', linestyle=':', linewidth=2, alpha=0.8)
    ax2.axhline(eq['w_inf'], color='tab:green', linestyle=':', linewidth=2, alpha=0.8)
    ax2.axhline(eq['w_fallback'], color='purple', linestyle=':', linewidth=2, alpha=0.8)
    ax2.plot([], [], color='gray', linestyle=':', linewidth=2, label='Analytical Equilibrium')

ax2.set_xlabel('Simulation Period', fontsize=11)
ax2.set_ylabel('Wage Level', fontsize=11)
ax2.set_title('Evolution of Wages over Time', fontsize=13)
ax2.legend(fontsize=10, loc='upper right')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

`get_sim_mean(metric_key, n_periods)` calculates the average of the last `n_periods` values in a time series. It is used to estimate the empirical steady state of the ABM across different time windows and compare it with the analytical equilibrium.

```python
def get_sim_mean(metric_key, n_periods):
    return np.mean(history[metric_key][-n_periods:]) if len(history[metric_key]) > 0 else 0
```

## Simulation with sudden shocks

### Unemployement Benefit Shock

The counterfactual experiment is defined as follows: a permanent shock that increases `w_bar` from 0.05 to 0.15 (a 200% increase in the unemployment benefit) at time `t=200` in the post-equilibrium simulation. Economically, this simulates an unexpected expansion of unemployment insurance. The system is initialised from the final state of the main simulation, ensuring that the shock starts from equilibrium.

```python
T_shock = 2000
t_shock_time = 200
shock_w_bar = 0.15
```

`worker_state` and `params` are copied so as not to alter the main simulation. This separation allows the trajectory to be compared with and without a shock, whilst maintaining the same starting point.

```python
shock_state = worker_state.copy()
p_shock = params.copy()
```

```python
history_shock = {
    'formal': [],
    'informal': [],
    'unemployed': [],
    'formal_wage': [],
    'informal_wage': [],
    'fallback_wage': [],
    'w_bar_level': []
}
```

The loop runs for 2000 periods. At `t=200`, the shock `w_bar -> 0.20` is applied. The metrics are stored in `history_shock`. Economically, the increase in `w_bar` raises the fallback wage `w_fallback`, which increases the NSC, reduces formal employment (the firm cannot afford the new efficiency wage) and leads to a shift towards informality and unemployment.

```python
for t in range(T_shock):
    if t == t_shock_time:
        p_shock['w_bar'] = shock_w_bar
        
    formal_mask = shock_state > 0
    informal_mask = shock_state == -1
    unemp_mask = shock_state == 0
    
    L_f = np.mean(formal_mask)
    L_i = np.mean(informal_mask)
    u_pure = np.mean(unemp_mask)
    non_formal_total = L_i + u_pure
    
    current_w_inf = max(p_shock['w_bar'], p_shock['w_inf_max'] - p_shock['gamma_inf'] * L_i)
    
    if non_formal_total > 0:
        w_fallback = (u_pure * p_shock['w_bar'] + L_i * current_w_inf) / non_formal_total
    else:
        w_fallback = p_shock['w_bar']
        
    j_employed = shock_state[formal_mask] - 1
    if len(j_employed) > 0:
        wages_i = np.array([nsc_wage(L_f, {**p_shock, 'q': q_firms[j]}, w_fallback) for j in j_employed])
        avg_formal_wage = wages_i.mean()
    else:
        avg_formal_wage = np.nan
        
    history_shock['formal'].append(L_f)
    history_shock['informal'].append(L_i)
    history_shock['unemployed'].append(u_pure)
    history_shock['formal_wage'].append(avg_formal_wage if not np.isnan(avg_formal_wage) else 0)
    history_shock['informal_wage'].append(current_w_inf)
    history_shock['fallback_wage'].append(w_fallback)
    history_shock['w_bar_level'].append(p_shock['w_bar'])
    
    new_state = shock_state.copy()
    
    emp_idx = np.where(formal_mask)[0]
    if len(emp_idx) > 0:
        sep = np.random.random(len(emp_idx)) < (p_shock['b'] * dt)
        new_state[emp_idx[sep]] = 0
        
    inf_idx = np.where(informal_mask)[0]
    if len(inf_idx) > 0:
        inf_sep = np.random.random(len(inf_idx)) < (p_shock['delta_i'] * dt)
        new_state[inf_idx[inf_sep]] = 0
        
    u_idx = np.where(unemp_mask)[0]
    if len(u_idx) > 0:
        gets_inf = np.random.random(len(u_idx)) < (p_shock['lambda_i'] * dt)
        new_state[u_idx[gets_inf]] = -1
        
    avail_idx = np.where(~formal_mask)[0] 
    if len(avail_idx) > 0 and non_formal_total > 0:
        current_mpl = labor_demand_mpl(L_f, p_shock)
        wage_gap = current_mpl - avg_formal_wage if not np.isnan(avg_formal_wage) else 0
        
        a_base = p_shock['b'] * L_f / max(non_formal_total, 0.001)  
        a = a_base * (1.0 + 3.0 * wage_gap) 
        a = max(0.0, a)  
        
        a_dt = min(a * dt, 0.99)
        gets_formal = np.random.random(len(avail_idx)) < a_dt
        hired = avail_idx[gets_formal]
        
        if len(hired) > 0:
            new_state[hired] = np.random.randint(1, N_firms + 1, len(hired))
            
    shock_state = new_state
```

Two charts are presented: (1) the trend in the shares of formal and informal employment and unemployment before and after the shock, and (2) the dynamics of formal, informal and fallback wages. The dotted vertical line indicates the point at which the shock occurred. The chart also shows the proportion of total employment accounted for by each sector (formal vs informal), excluding unemployment.

```python
fig2, (ax3, ax4, ax_share_wbar) = plt.subplots(3, 1, figsize=(10, 14))

time_axis_shock = range(T_shock)
ax3.plot(time_axis_shock, history_shock['formal'], color='steelblue', linewidth=2, label='Formal Employment ($L_f$)')
ax3.plot(time_axis_shock, history_shock['informal'], color='tab:green', linewidth=2, label='Informal Employment ($L_{inf}$)')
ax3.plot(time_axis_shock, history_shock['unemployed'], color='firebrick', linewidth=2, label='Pure Unemployment ($U$)')

ax3.fill_between(time_axis_shock, 0, history_shock['formal'], color='steelblue', alpha=0.1)
ax3.fill_between(time_axis_shock, history_shock['formal'], np.array(history_shock['formal']) + np.array(history_shock['informal']), color='tab:green', alpha=0.1)

ax3.axvline(t_shock_time, color='black', linestyle='--', linewidth=2, label='Shock: $w_{bar}$ increases')

ax3.set_xlabel('Simulation Period (Post-Equilibrium)', fontsize=11)
ax3.set_ylabel('Population Fraction', fontsize=11)
ax3.set_title('Labor Market Response to an Unemployment Benefit Shock', fontsize=13)
ax3.legend(fontsize=10, loc='center right')
ax3.grid(True, alpha=0.3)
ax3.set_ylim(0, 1)

ax4.plot(time_axis_shock, history_shock['formal_wage'], color='steelblue', linewidth=2, label='Avg Formal Wage ($w^*$ NSC)')
ax4.plot(time_axis_shock, history_shock['informal_wage'], color='tab:green', linewidth=2, label='Informal Wage ($w_{inf}$)')
ax4.plot(time_axis_shock, history_shock['fallback_wage'], color='purple', linestyle='--', linewidth=2, label='Expected Fallback Wage ($w_{fallback}$)')
ax4.plot(time_axis_shock, history_shock['w_bar_level'], color='firebrick', linestyle=':', linewidth=2, label='Unemployment Benefit ($w_{bar}$)')

ax4.axvline(t_shock_time, color='black', linestyle='--', linewidth=2)

ax4.set_xlabel('Simulation Period (Post-Equilibrium)', fontsize=11)
ax4.set_ylabel('Wage Level', fontsize=11)
ax4.set_title('Wage Dynamics Response to Shock', fontsize=13)
ax4.legend(fontsize=10, loc='upper right')
ax4.grid(True, alpha=0.3)


total_emp_shock = [f + i for f, i in zip(history_shock['formal'], history_shock['informal'])]
share_formal_shock = [f/t if t > 0 else 0 for f, t in zip(history_shock['formal'], total_emp_shock)]
share_informal_shock = [i/t if t > 0 else 0 for i, t in zip(history_shock['informal'], total_emp_shock)]

ax_share_wbar.stackplot(time_axis_shock,
    share_formal_shock, share_informal_shock,
    labels=['Formal Share', 'Informal Share'],
    colors=['steelblue', 'tab:green'], alpha=0.7)
ax_share_wbar.axvline(t_shock_time, color='black', linestyle='--', linewidth=2, label='Shock: $w_{bar}$ increases')
ax_share_wbar.set_xlabel('Simulation Period (Post-Equilibrium)', fontsize=11)
ax_share_wbar.set_ylabel('Share of Total Employment', fontsize=11)
ax_share_wbar.set_title('Employment Share by Sector (of Total Employed) - $w_{bar}$ Shock', fontsize=13)
ax_share_wbar.legend(fontsize=10, loc='upper right')
ax_share_wbar.grid(True, alpha=0.3)
ax_share_wbar.set_ylim(0, 1)

plt.tight_layout()
plt.show()
```

### Various shocks at the same time: AI in the labor market

One can imagine the AI revolution having an effect in more than only one parameter. This makes hard to predict what's going to happend in the future unless one uses this models and simulate the case.

We expect that a shock as AI can affect the following parameters:

- Productivity: We expect productivity to increase since now people can use AI as a complement (some argue that it may be a substitute as well)
- Monitoring: We expect better monitoring to human work. In other words, it might become easier for business to check if the workers are putting in effort.
- Separation rate: We expect more people to loose their job due to exogenous reasons. In this case, the resaons would be AI replacement.
- Informal job adquisition rate: WE expect that finding informal jobs becomes harder. Many of the known informal jobs can now be done with AI which may be cheaper and better at it.

This conclusions are driven by the assumptions we have on the effect that AI will have on the labor market and business, any change on parameters would make the market react in a different way.

It is also important to say that the assumptions of the models also afect how it reacts to this changes. For instance, under the way the model is built, business depend on labor to produce (given their production function) while they don't have dependencies in capital. One can expect that AI is seen as capital and it can complement or even substitute the labor. This is left out of the model so far.

We believe that this are the most plausible shocks based on the book Prediction machines: the simple economics of artificial intelligence from Ajay Agrawal, Joshua Gans and Avi Goldfarb. This book introduces the idea of AI as jsut a prediction improver, based on this definition we assumed the effects on the economy as the ones we present. It is true that this is a strong assumption and we welcome to using different changes in the parameters. We believe that this is an excercise to check what would happend based on out model but this is highly dependent on the magnitude of the changes. In a more professional setup it would be necessary to calibrate them or give stronger arguments of why we expect such change and why in that magnitude.

```python
T_shock_Multi = 2000
t_shock_time_Multi = 200

shock_multi_A = 0.5          
shock_multi_q = 0.8          
shock_multi_b = 0.5         
shock_multi_lambda_i = 0.08 

shock_state_multi = worker_state.copy()
p_shock_multi = params.copy()

q_firms_multi = q_firms.copy()

history_shock_multi = {
    'formal': [], 'informal': [], 'unemployed': [],
    'formal_wage': [], 'informal_wage': [], 'fallback_wage': []
}

for t in range(T_shock_Multi):
    if t == t_shock_time_Multi:
        p_shock_multi['A'] = shock_multi_A
        p_shock_multi['q'] = shock_multi_q
        p_shock_multi['b'] = shock_multi_b
        p_shock_multi['lambda_i'] = shock_multi_lambda_i
        
        q_firms_multi = np.clip(np.random.normal(loc=shock_multi_q, scale=0.05, size=N_firms), 0.05, 0.95)

    formal_mask = shock_state_multi > 0
    informal_mask = shock_state_multi == -1
    unemp_mask = shock_state_multi == 0
    
    L_f = np.mean(formal_mask)
    L_i = np.mean(informal_mask)
    u_pure = np.mean(unemp_mask)
    non_formal_total = L_i + u_pure
    
    current_w_inf = max(p_shock_multi['w_bar'], p_shock_multi['w_inf_max'] - p_shock_multi['gamma_inf'] * L_i)
    
    if non_formal_total > 0:
        w_fallback = (u_pure * p_shock_multi['w_bar'] + L_i * current_w_inf) / non_formal_total
    else:
        w_fallback = p_shock_multi['w_bar']
        
    j_employed = shock_state_multi[formal_mask] - 1
    if len(j_employed) > 0:
        wages_i = np.array([nsc_wage(L_f, {**p_shock_multi, 'q': q_firms_multi[j]}, w_fallback) for j in j_employed])
        avg_formal_wage = wages_i.mean()
    else:
        avg_formal_wage = np.nan
        
    history_shock_multi['formal'].append(L_f)
    history_shock_multi['informal'].append(L_i)
    history_shock_multi['unemployed'].append(u_pure)
    history_shock_multi['formal_wage'].append(avg_formal_wage if not np.isnan(avg_formal_wage) else 0)
    history_shock_multi['informal_wage'].append(current_w_inf)
    history_shock_multi['fallback_wage'].append(w_fallback)
    
    new_state = shock_state_multi.copy()
    
    emp_idx = np.where(formal_mask)[0]
    if len(emp_idx) > 0:
        sep = np.random.random(len(emp_idx)) < (p_shock_multi['b'] * dt)
        new_state[emp_idx[sep]] = 0
        
    inf_idx = np.where(informal_mask)[0]
    if len(inf_idx) > 0:
        inf_sep = np.random.random(len(inf_idx)) < (p_shock_multi['delta_i'] * dt)
        new_state[inf_idx[inf_sep]] = 0
        
    u_idx = np.where(unemp_mask)[0]
    if len(u_idx) > 0:
        gets_inf = np.random.random(len(u_idx)) < (p_shock_multi['lambda_i'] * dt)
        new_state[u_idx[gets_inf]] = -1
        
    avail_idx = np.where(~formal_mask)[0] 
    if len(avail_idx) > 0 and non_formal_total > 0:
        current_mpl = labor_demand_mpl(L_f, p_shock_multi)
        wage_gap = current_mpl - avg_formal_wage if not np.isnan(avg_formal_wage) else 0
        
        a_base = p_shock_multi['b'] * L_f / max(non_formal_total, 0.001)  
        a = a_base * (1.0 + 3.0 * wage_gap) 
        a = max(0.0, a)  
        
        a_dt = min(a * dt, 0.99)
        gets_formal = np.random.random(len(avail_idx)) < a_dt
        hired = avail_idx[gets_formal]
        
        if len(hired) > 0:
            new_state[hired] = np.random.randint(1, N_firms + 1, len(hired))
            
    shock_state_multi = new_state
```

### Can our model predict productivity?

Now we will check if our model would work in order to check the productivity. We want to see if by using the unemployement and informality rate we can get information about productivity. 

To do so, we assume that first the model is actually right and this is the way labor market works and it shows how the world works. Second, we assume that the others parameters are held constant throught time. 

We now create a copy of the dataframe only with Colombia

```python
df_col = df_pivot[df_pivot['Country Name'] == 'Colombia'].copy()
df_col = df_col.sort_values('Year')
```

```python
years_list = []
real_f, real_i, real_u = [], [], []
mod_f, mod_i, mod_u = [], [], []
calibrated_A = []
```

We run this loop year by year through the data. We pull the actual percentages of formal, informal, and unemployed workers to use as our ground truth. We then define a goal (the objective function). 
It takes a guess at a Productivity ($A$) and Informal Search Efficiency ($\lambda_i$) might be. Use those guesses to run the model, and then calculate the gap between the model's results and the real-world data.

We then look for the best possible values. It starts with your default settings and tweaks them within the boundaries you set—for example, making sure the search rate stays between 1% and 95%. It keeps adjusting these two knobs until the Sum of Squared Errors is as small as it can possibly be.

We then check if the optimizer actually found a solution. If it did, we save those parameters and the model’s version of the labor stats.

This gives you a play-by-play of the specific Productivity and Search Speed values that best explain the labor market conditions for that specific point in time.

```python
years_list = []
real_f, real_i, real_u = [], [], []
mod_f, mod_i, mod_u = [], [], []
calibrated_A = []
calibrated_lambda_i = [] 

for index, row in df_col.iterrows():
    year = int(row['Year'])
    emp_formal = row['% Formal'] / 100
    emp_informal = row['% Informal'] / 100
    emp_unemployed = row['% Unemployed'] / 100
    
    def objective_year(x):
        p_cal = params.copy()
        p_cal['A'] = x[0]          
        p_cal['lambda_i'] = x[1]   
        
        eq_cal = find_analytical_eq(p_cal)
        
        if eq_cal is None:
            return 1e6 
            
        error = (eq_cal['L_f'] - emp_formal)**2 + \
                (eq_cal['L_i'] - emp_informal)**2 + \
                (eq_cal['U'] - emp_unemployed)**2
        return error

    x0 = [params['A'], params['lambda_i']]
    
    bounds = [(0.1, 10.0), (0.01, 0.95)]
    
    res = minimize(objective_year, x0, bounds=bounds, method='L-BFGS-B')
    
    if res.success:
        p_opt = params.copy()
        p_opt['A'] = res.x[0]
        p_opt['lambda_i'] = res.x[1]
        
        eq_opt = find_analytical_eq(p_opt)
        
        years_list.append(year)
        real_f.append(emp_formal)
        real_i.append(emp_informal)
        real_u.append(emp_unemployed)
        
        mod_f.append(eq_opt['L_f'])
        mod_i.append(eq_opt['L_i'])
        mod_u.append(eq_opt['U'])
        
        calibrated_A.append(p_opt['A'])
        calibrated_lambda_i.append(p_opt['lambda_i'])
```

## 8. Discussion and conclusion

## Summary of Findings and Synthesis

This research sought to evaluate the structural integrity of the Shapiro-Stiglitz (1984) efficiency wage framework when extended to account for agent heterogeneity and a dual labor market structure. By utilizing both analytical root-finding and a stochastic Agent-Based Model (ABM), we have identified several key departures from the canonical "Work vs. Pure Unemployment" binary.

**Core Theoretical Insight: The Endogenous Fallback Mechanism**
The primary finding is that the informal sector acts as a dynamic fallback option, which fundamentally alters the worker’s intertemporal optimization problem. In the standard model, the opportunity cost of job loss is tied to a static unemployment benefit ($\bar{w}$). Our model endogenizes this cost: as the informal sector ($L_{inf}$) becomes more productive, the reservation utility of formal workers rises. From an efficiency wage perspective, formal firms are forced to offer a higher wage premium to maintain the requisite disciplinary "penalty" for shirking.

**Welfare Dynamics and the "Tipping Point" of Informality**
By comparing the total social surplus of the dual-sector economy against the standard model, we answered a critical question: *Does the informal sector act as a safety net or a parasite on formal growth?* The analysis reveals a distinct "Tipping Point":
*   **Destructive Informality (The Crowding-Out Penalty):** When the informal sector is moderately productive and highly congested, it acts as a drag on the economy. It forces formal firms to raise wages and fire workers (the Discipline Effect), but those displaced workers produce significantly less value in the informal sector than they did in formal factories. The formal deadweight loss exceeds the informal output gain, resulting in a net negative welfare effect.
*   **Productive Informality (The Gig-Economy Premium):** Conversely, if a technological shock (e.g., app-based gig work) raises the maximum informal productivity ($w_{inf\_max}$), the informal sector becomes an engine of growth. The direct output of these new informal jobs mathematically overwhelms the deadweight loss of formal unemployment, leading to a net positive welfare premium.

**Agent-Based Convergence and Heterogeneity**
A significant technical finding was the robustness of the "Worker Discipline Device" under micro-level noise. Even with heterogeneous discount rates ($r_i$) and idiosyncratic monitoring intensities ($q_j$), the aggregate labor market effectively "averages out" these differences to converge toward the analytical steady state.

## Empirical performance and the productivity gap

The calibration of the model to Colombian macroeconomic data (2000–2023) serves as a diagnostic test of the theory’s explanatory power. While the model successfully "unmasked" the directional impact of the 2020 COVID-19 shock—attributing the collapse in formal employment to a drastic reduction in implicit productivity ($A$)—the divergences noted in the comparison with official DANE data are telling.

The model’s "Implicit Productivity" exhibited higher volatility than official metrics. We posit that this is due to the model forcing all labor market adjustments through the lens of productivity ($A$) and search efficiency ($\lambda_i$). In reality, the 2011–2015 "decoupling" likely reflects the influence of omitted variables, such as capital-intensive investments in extractive industries and shifts in the terms of trade, which our labor-only production function cannot capture.

## Methodological Weaknesses and Future Research

While the extended Shapiro-Stiglitz model provides a rigorous framework for understanding dual economies, our welfare analysis revealed several structural limitations that provide a roadmap for future econometric and theoretical inquiry:

**1. The "Zero Effort" Bias in Informal Welfare ($e_{inf} = 0$)**
In our welfare calculation, the disutility of effort ($e$) was strictly subtracted for formal workers, but no equivalent cost was applied to informal labor. This heavily biased the model toward finding a positive "Informal Premium," as society mathematically "saved" effort costs every time a formal worker was fired and moved to the shadow economy. Future research must incorporate an informal disutility parameter ($e_{inf}$) and a risk penalty ($\rho$) to reflect the precarious, physically demanding reality of informal labor.

**2. Omission of the Fiscal and Tax Wedge ($\tau$)**
Our model assumes the unemployment benefit ($\bar{w}$) is exogenously funded. In reality, formal firms pay payroll taxes ($\tau$) to fund these benefits, while the informal sector evades them. A better model would endogenize the government budget constraint ($\tau \cdot w_f \cdot L_f = \bar{w} \cdot U$). Under this mechanism, as informality grows, taxes on the shrinking formal sector must rise, creating a vicious cycle of formal job destruction that accurately captures the societal penalty of tax evasion.

**3. Omission of Capital ($K$) and Idle Machinery Costs**
The current production function, $Y = A \cdot L^\alpha$, assumes labor is the sole factor of production. When efficiency wages rise and formal employment drops, the true macroeconomic cost includes the lost productivity of idle capital (machinery, software, infrastructure). Future research should incorporate a Capital-Augmented Cobb-Douglas function ($Y_f = A \cdot K^{1-\alpha} \cdot L_f^\alpha$) so that formal job destruction triggers a realistic collapse in total output.

**4. Parameter Endogeneity (The Monitoring Choice)**
We treated monitoring intensity ($q$) as an exogenous constant. In a more complete general equilibrium framework, $q$ should be an endogenous choice for firms, who face a trade-off between investing in monitoring technology (e.g., AI surveillance) and paying higher efficiency wages.

In conclusion, while this dual-sector extension proves that efficiency wages and endogenous outside options are vital for understanding emerging labor markets, it remains a "first-approximation" tool. Integrating fiscal constraints, capital depreciation, and true informal effort costs is the necessary next step to fully capture the complexity of globalized macroeconomics.
